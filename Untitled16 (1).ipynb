```python
text = input("Enter text: ").lower()
for ch in ".,!?;:'\"()[]{}":
    text = text.replace(ch, "")
words = text.split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1
for word, count in freq.items():
    print(f"{word}: {count}")
```

## F

```python
inventory = {}

def add_product(name, qty, price):
    inventory[name] = {"quantity": qty, "price": price}

def update_quantity(name, qty):
    if name in inventory:
        inventory[name]["quantity"] += qty

def total_value():
    return sum(item["quantity"] * item["price"] for item in inventory.values())

def low_stock_report(threshold=5):
    return {name: info for name, info in inventory.items() if info["quantity"] < threshold}
```

## G

```python
grades = {
    "Alice": {"Math": 85, "English": 90, "Science": 88},
    "Bob": {"Math": 78, "English": 85, "Science": 92},
    "Charlie": {"Math": 92, "English": 88, "Science": 95}
}

def calculate_average(grades):
    return {name: sum(subjects.values()) / len(subjects) for name, subjects in grades.items()}

def find_subject_grades(grades, subject):
    subject_grades = {name: subjects[subject] for name, subjects in grades.items() if subject in subjects}
    return max(subject_grades, key=subject_grades.get), min(subject_grades, key=subject_grades.get)

def sort_by_average(grades):
    averages = calculate_average(grades)
    return sorted(averages.items(), key=lambda x: x[1], reverse=True)
```

```python
products = {
    "apple": {"price": 1.0, "quantity": 100},
    "banana": {"price": 0.5, "quantity": 50},
    "orange": {"price": 0.8, "quantity": 75}
}

cart = {}

def add_to_cart(product, qty):
    if products[product]["quantity"] >= qty:
        cart[product] = cart.get(product, 0) + qty
        products[product]["quantity"] -= qty

def remove_from_cart(product, qty):
    if cart.get(product, 0) >= qty:
        cart[product] -= qty
        products[product]["quantity"] += qty

def calculate_total():
    return sum(products[product]["price"] * qty for product, qty in cart.items())

def apply_discount_and_tax(total, discount=0, tax_rate=0.1):
    return total * (1 - discount) * (1 + tax_rate)
```
